```python
df = pd.read_csv("cleaned_sales_data (1).csv", parse_dates=["Date"])
df.head()
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Order IDs:", df["Order_ID"].duplicated().sum())
df = df.drop_duplicates()
```

```python
# Total_Sales = Quantity * Unit_Price
print("Sales calculation correct:", (df["Quantity"] * df["Unit_Price"] == df["Total_Sales"]).all())

# negative leda zero values levu ani check
print("No negative values:", (df[["Quantity", "Unit_Price", "Total_Sales"]] > 0).all().all())

# text columns lo extra spaces teesey
for col in ["Customer", "City", "Product", "Category"]:
    df[col] = df[col].str.strip()

# dates range
print("Date range:", df["Date"].min(), "to", df["Date"].max())

# unique values chudadam
print(df["City"].unique())
print(df["Category"].unique())
```

```python
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Total_Sales"] < lower) | (df["Total_Sales"] > upper)]
print("Lower limit:", lower, "Upper limit:", upper)
print("Outliers:", len(outliers))

# box plot
df["Total_Sales"].plot(kind="box")
plt.title("Total Sales - Outlier Check")
plt.show()
```

```python
print(df[["Quantity", "Unit_Price", "Total_Sales"]].describe())
print("Median sales:", df["Total_Sales"].median())
print("Std deviation:", round(df["Total_Sales"].std(), 2))
```

```python
corr = df[["Quantity", "Unit_Price", "Total_Sales"]].corr()
print(corr)

plt.imshow(corr, cmap="Blues")
plt.colorbar()
plt.xticks(range(3), corr.columns, rotation=45)
plt.yticks(range(3), corr.columns)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
daily = df.groupby("Date")["Total_Sales"].sum()
daily.index = daily.index.strftime("%d %b")
daily.plot(kind="line", marker="o")
plt.title("Daily Sales Trend")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# product wise revenue vs quantity
print(df.groupby("Product")[["Quantity", "Total_Sales"]].sum().sort_values("Total_Sales", ascending=False))
```
